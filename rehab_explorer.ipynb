```python
label = np.load('Rehab_assessment/Label.npy', allow_pickle=True)
fma = label.sum(axis=1)

print("mean FMA lines 0–119 (hipotesis: baseline):", fma[:120].mean().round(2))
print("mean FMA linhas 120–221 (hipotesis: follow-up):", fma[120:].mean().round(2))
print("\nTotal assessments:", len(fma))
```

```python
# Distribuição dos scores totais
print("=== Total score (27 itens sum, max=54) ===")
print(f"Min: {fma.min()}, Max: {fma.max()}, Mean: {fma.mean():.1f}, Median: {np.median(fma):.1f}")

# Item score (columns)
print("\n=== Zeros per item (% assessments where the patient couldn't do) ===")
for i in range(27):
    zeros = (label[:, i] == 0).sum()
    print(f"Item {i:02d}: {zeros}/222 zeros ({zeros/222*100:.0f}%)")
```

```python
# Chanel values
sig = np.load('Rehab_assessment/d02_processed_data/001/000_1.npy', allow_pickle=True)

print("Shape:", sig.shape)
print("\nChanel stats:")
print(f"{'Chanel':>6}  {'Min':>8}  {'Max':>8}  {'Std':>8}")
for i in range(12):
    print(f"  [{i:2d}]  {sig[:,i].min():8.2f}  {sig[:,i].max():8.2f}  {sig[:,i].std():8.2f}")
```

```python
# Movement 0 = shoulder elevation (proximal) vs. movement 20 = finger flexion (distal)
prox = np.load('Rehab_assessment/d02_processed_data/001/000_1.npy', allow_pickle=True)
dist = np.load('Rehab_assessment/d02_processed_data/001/020_1.npy', allow_pickle=True)

print("Std chanel — shoulder elevation:")
print([round(prox[:,i].std(), 2) for i in range(12)])

print("\nStd chanel — finger flexion:")
print([round(dist[:,i].std(), 2) for i in range(12)])
```

```python
results = []
for mov in range(27):
    file_exists = 0
    signal_nonzero = 0
    for folder in range(1, 223):
        path = f'Rehab_assessment/d02_processed_data/{folder:03d}/{mov:03d}_1.npy'
        if os.path.exists(path):
            file_exists += 1
            s = np.load(path, allow_pickle=True)
            if s.sum() != 0:
                signal_nonzero += 1
    results.append((mov, file_exists, signal_nonzero))

print(f"{'Mov':>4}  {'File':>8}  {'Signal≠0':>8}  {'% util':>8}")
for mov, exists, nonzero in results:
    print(f"  {mov:02d}  {exists:8d}  {nonzero:8d}  {nonzero/222*100:7.1f}%")
```
