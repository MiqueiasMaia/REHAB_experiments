```python
import numpy as np
import os
```

```python
for root, dirs, files in os.walk('Rehab_assessment'):
    level = root.replace('Rehab_assessment', '').count(os.sep)
    indent = '  ' * level
    
    print(f'{indent}{os.path.basename(root)}/')
    
    if level < 2: 
        for f in files:
            print(f'{indent}  {f}')
```
